```python
# === KOMÓRKA 3: GŁÓWNY KOD APLIKACJI ===
import sqlite3
import threading
from flask import Flask, request, redirect, url_for, session, render_template_string
from werkzeug.security import generate_password_hash, check_password_hash

app = Flask(__name__)
app.secret_key = 'medica_plus_secret_key_2025'
DB_PATH = 'medica.db'

# ---------------------------------------------------------------------
# 1. INICJALIZACJA BAZY DANYCH (SQLite - baza wbudowana)
# ---------------------------------------------------------------------
def init_db():
    conn = sqlite3.connect(DB_PATH)
    conn.execute('''
        CREATE TABLE IF NOT EXISTS pacjenci (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            imie TEXT NOT NULL,
            nazwisko TEXT NOT NULL,
            pesel TEXT UNIQUE NOT NULL,
            data_urodzenia TEXT NOT NULL,
            telefon TEXT,
            email TEXT,
            login TEXT UNIQUE NOT NULL,
            haslo_hash TEXT NOT NULL,
            data_rejestracji TEXT DEFAULT CURRENT_TIMESTAMP
        )
    ''')
    conn.commit()
    conn.close()

init_db()


# ---------------------------------------------------------------------
# 2. SZABLONY HTML (CSS w stylu Medica+)
# ---------------------------------------------------------------------
BASE_CSS = '''
<style>
    * { box-sizing: border-box; margin: 0; padding: 0; }
    body { font-family: 'Segoe UI', Arial, sans-serif; background: #f0f4f8;
           color: #2c3e50; padding: 30px; min-height: 100vh; }
    .container { max-width: 650px; margin: 0 auto; background: white;
                 padding: 35px; border-radius: 12px;
                 box-shadow: 0 4px 20px rgba(0,0,0,0.08); }
    h1 { color: #0d7377; margin-bottom: 8px; font-size: 26px; }
    .subtitle { color: #7f8c8d; margin-bottom: 25px; font-size: 14px; }
    .logo { background: #0d7377; color: white; width: 60px; height: 60px;
            border-radius: 50%; display: flex; align-items: center;
            justify-content: center; font-size: 28px; font-weight: bold;
            margin-bottom: 15px; }
    label { display: block; margin-top: 14px; margin-bottom: 5px;
            font-weight: 600; color: #34495e; font-size: 14px; }
    input { width: 100%; padding: 10px 12px; border: 1px solid #dce4ec;
            border-radius: 6px; font-size: 14px; }
    input:focus { outline: none; border-color: #0d7377; }
    button { background: #0d7377; color: white; padding: 12px 28px; border: none;
             border-radius: 6px; cursor: pointer; font-size: 15px; font-weight: 600;
             margin-top: 20px; }
    button:hover { background: #0a5a5d; }
    .nav { margin-bottom: 20px; padding-bottom: 15px; border-bottom: 1px solid #ecf0f1; }
    .nav a { color: #0d7377; text-decoration: none; margin-right: 20px;
             font-weight: 500; font-size: 14px; }
    .alert { padding: 12px 16px; border-radius: 6px; margin-bottom: 20px; font-size: 14px; }
    .alert-success { background: #d4edda; color: #155724; border: 1px solid #c3e6cb; }
    .alert-error { background: #f8d7da; color: #721c24; border: 1px solid #f5c6cb; }
    .card-row { display: flex; padding: 12px 0; border-bottom: 1px solid #ecf0f1; }
    .card-label { font-weight: 600; color: #7f8c8d; width: 180px; }
    .card-value { color: #2c3e50; flex: 1; }
    .row { display: flex; gap: 15px; }
    .row > div { flex: 1; }
</style>
'''

TEMPLATE_INDEX = BASE_CSS + '''
<div class="container">
    <div class="logo">M+</div>
    <h1>Medica+ — System Obsługi Pacjentów</h1>
    <p class="subtitle">Sieć klinik ambulatoryjnych</p>
    <div class="nav">
        <a href="/dodaj">➕ Rejestracja pacjenta</a>
        <a href="/logowanie">🔑 Logowanie</a>
    </div>
    <p>Witamy w systemie obsługi pacjentów Medica+. Wybierz opcję powyżej.</p>
</div>
'''

TEMPLATE_DODAJ = BASE_CSS + '''
<div class="container">
    <div class="nav"><a href="/">← Strona główna</a></div>
    <h1>Rejestracja nowego pacjenta</h1>
    <p class="subtitle">Wypełnij formularz, aby założyć konto</p>
    {% if msg %}<div class="alert alert-{{msg_type}}">{{ msg }}</div>{% endif %}
    <form method="POST">
        <div class="row">
            <div><label>Imię *</label><input name="imie" required></div>
            <div><label>Nazwisko *</label><input name="nazwisko" required></div>
        </div>
        <div class="row">
            <div><label>PESEL *</label><input name="pesel" required maxlength="11"></div>
            <div><label>Data urodzenia *</label><input name="data_urodzenia" type="date" required></div>
        </div>
        <div class="row">
            <div><label>Telefon</label><input name="telefon" type="tel"></div>
            <div><label>Email</label><input name="email" type="email"></div>
        </div>
        <div class="row">
            <div><label>Login *</label><input name="login" required></div>
            <div><label>Hasło *</label><input name="haslo" type="password" required></div>
        </div>
        <button type="submit">Zarejestruj pacjenta</button>
    </form>
</div>
'''

TEMPLATE_LOGIN = BASE_CSS + '''
<div class="container">
    <div class="nav"><a href="/">← Strona główna</a></div>
    <h1>Logowanie pacjenta</h1>
    <p class="subtitle">Zaloguj się, aby zobaczyć swoją kartę pacjenta</p>
    {% if msg %}<div class="alert alert-{{msg_type}}">{{ msg }}</div>{% endif %}
    <form method="POST">
        <label>Login</label>
        <input name="login" required autofocus>
        <label>Hasło</label>
        <input name="haslo" type="password" required>
        <button type="submit">Zaloguj się</button>
    </form>
</div>
'''

TEMPLATE_KARTA = BASE_CSS + '''
<div class="container">
    <div class="nav">
        <a href="/">← Strona główna</a>
        <a href="/wyloguj">🚪 Wyloguj</a>
    </div>
    <h1>Karta pacjenta</h1>
    <p class="subtitle">Dane osobowe i kontaktowe</p>
    <div class="alert alert-success">Witaj, {{ p['imie'] }} {{ p['nazwisko'] }}!</div>
    <div class="card-row"><div class="card-label">ID pacjenta</div><div class="card-value">#{{ p['id'] }}</div></div>
    <div class="card-row"><div class="card-label">Imię i nazwisko</div><div class="card-value">{{ p['imie'] }} {{ p['nazwisko'] }}</div></div>
    <div class="card-row"><div class="card-label">PESEL</div><div class="card-value">{{ p['pesel'] }}</div></div>
    <div class="card-row"><div class="card-label">Data urodzenia</div><div class="card-value">{{ p['data_urodzenia'] }}</div></div>
    <div class="card-row"><div class="card-label">Telefon</div><div class="card-value">{{ p['telefon'] or '—' }}</div></div>
    <div class="card-row"><div class="card-label">Email</div><div class="card-value">{{ p['email'] or '—' }}</div></div>
    <div class="card-row"><div class="card-label">Login</div><div class="card-value">{{ p['login'] }}</div></div>
    <div class="card-row"><div class="card-label">Data rejestracji</div><div class="card-value">{{ p['data_rejestracji'] }}</div></div>
</div>
'''


# ---------------------------------------------------------------------
# 3. TRASY (ENDPOINTY) APLIKACJI
# ---------------------------------------------------------------------
@app.route('/')
def index():
    return render_template_string(TEMPLATE_INDEX)


@app.route('/dodaj', methods=['GET', 'POST'])
def dodaj_pacjenta():
    msg, msg_type = None, None
    if request.method == 'POST':
        try:
            conn = sqlite3.connect(DB_PATH)
            conn.execute('''
                INSERT INTO pacjenci (imie, nazwisko, pesel, data_urodzenia,
                    telefon, email, login, haslo_hash)
                VALUES (?, ?, ?, ?, ?, ?, ?, ?)
            ''', (
                request.form['imie'], request.form['nazwisko'],
                request.form['pesel'], request.form['data_urodzenia'],
                request.form.get('telefon', ''), request.form.get('email', ''),
                request.form['login'],
                generate_password_hash(request.form['haslo'])
            ))
            conn.commit()
            conn.close()
            msg = f"✓ Pacjent {request.form['imie']} {request.form['nazwisko']} został pomyślnie zarejestrowany."
            msg_type = 'success'
        except sqlite3.IntegrityError:
            msg = "✗ Błąd: pacjent o podanym PESEL lub loginie już istnieje."
            msg_type = 'error'
        except Exception as e:
            msg = f"✗ Błąd: {str(e)}"
            msg_type = 'error'
    return render_template_string(TEMPLATE_DODAJ, msg=msg, msg_type=msg_type)


@app.route('/logowanie', methods=['GET', 'POST'])
def logowanie():
    msg, msg_type = None, None
    if request.method == 'POST':
        conn = sqlite3.connect(DB_PATH)
        conn.row_factory = sqlite3.Row
        cursor = conn.execute('SELECT * FROM pacjenci WHERE login = ?',
                              (request.form['login'],))
        pacjent = cursor.fetchone()
        conn.close()
        if pacjent and check_password_hash(pacjent['haslo_hash'],
                                           request.form['haslo']):
            session['pacjent_id'] = pacjent['id']
            return redirect(url_for('karta_pacjenta'))
        else:
            msg = "✗ Nieprawidłowy login lub hasło."
            msg_type = 'error'
    return render_template_string(TEMPLATE_LOGIN, msg=msg, msg_type=msg_type)


@app.route('/karta')
def karta_pacjenta():
    if 'pacjent_id' not in session:
        return redirect(url_for('logowanie'))
    conn = sqlite3.connect(DB_PATH)
    conn.row_factory = sqlite3.Row
    cursor = conn.execute('SELECT * FROM pacjenci WHERE id = ?',
                          (session['pacjent_id'],))
    pacjent = cursor.fetchone()
    conn.close()
    return render_template_string(TEMPLATE_KARTA, p=pacjent)


@app.route('/wyloguj')
def wyloguj():
    session.pop('pacjent_id', None)
    return redirect(url_for('index'))

```
